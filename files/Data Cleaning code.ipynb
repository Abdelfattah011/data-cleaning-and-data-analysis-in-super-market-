### Data Cleaning

#### 1.Rename Columns

```python
df.columns = df.columns.str.strip()
df.columns
```

```python
df.columns = df.columns.str.upper()
df.columns
```

```python
df.columns = df.columns.str.replace(' ', '_')
df.columns
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.to_csv('1. Renamed Data.csv', index=False)
```

#### 2. Handle missing values

```python
df["BRANCH"].value_counts()
```

```python
df["BRANCH"] = df["BRANCH"].str.replace("giza", "Giza")
df["BRANCH"] = df["BRANCH"].str.replace("cairo" , "Cairo")
df["BRANCH"] = df["BRANCH"].str.strip()
df["BRANCH"].value_counts()
```

```python
df["PAYMENT"].value_counts()
```

```python
df["PAYMENT"] = df["PAYMENT"].str.replace("cash", "Cash")
df["PAYMENT"] = df["PAYMENT"].str.replace("credit card", "Credit Card")
df["PAYMENT"] = df["PAYMENT"].str.replace("Credit card", "Credit Card")
df["PAYMENT"] = df["PAYMENT"].str.replace("Ewallet", "E-wallet")
df["PAYMENT"].value_counts()
```

```python
df["GENDER"] = df["GENDER"].replace("F", "Female")
df["GENDER"].value_counts()
df["GENDER"] = df["GENDER"].replace("M", "Male")
df["GENDER"].value_counts()
```

```python
df["DATE"] = df["DATE"].fillna("3/12/2019", inplace=True)
```

```python
df["DATE"].isnull().sum()
```

```python
df["DATE"].dtypes
```

```python
df["DATE"]
```

```python
df["DATE"] = pd.to_datetime(df["DATE"], errors="coerce")
df["DATE"].isnull().sum()
```

```python
df.isnull().sum()
```

```python
df
```

```python
df[df["UNIT_PRICE"].isnull()]
```

```python
df["UNIT_PRICE"] = df["UNIT_PRICE"].fillna(df["UNIT_PRICE"].mean(),)
df["UNIT_PRICE"].isnull().sum()
```

```python
df["SALES"] = df["SALES"].str.replace("EGP", "")
df["SALES"] = df["SALES"].astype(float)
df["SALES"] = df["SALES"].fillna(df["SALES"].mean())
df["SALES"].isnull().sum()
```

```python
df.isnull().sum()
```

#### 8. Create a new column called satisfied based on rating.

```python
def Satisfied(rate) :
  if rate >= 7.0:
    return("Satisfied")
  else:
    return("Not Satisfied")

df["SATISFITION"] = df["RATING"].apply(Satisfied)
df
```
